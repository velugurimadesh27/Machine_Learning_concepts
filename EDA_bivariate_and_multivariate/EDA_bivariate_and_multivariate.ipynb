```python
import pandas as pd
import seaborn as sns
```

```python
flights=sns.load_dataset('flights')
flights.head(3)
```

```python
tips=sns.load_dataset('tips')
tips.head(3)
```

```python
titanic=pd.read_csv('/content/Titanic-Dataset.csv')
titanic.head(3)
```

#Scatter plot(Numerical-Numerical)

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['smoker'],style=tips['sex'],size=tips['size'])
```

#BarPlot(Numerical-Categorical)

```python
sns.barplot(x=titanic['Pclass'],y=titanic['Age'],hue=titanic['Sex'])
```

#3.BoxPlot(Numerical-Categorical)

```python
sns.boxplot(x=titanic['Sex'],y=titanic['Age'],hue=titanic['Survived'])
```

#4.DistPlot(Numerical-Categorical)

```python
sns.displot(titanic[titanic['Survived']==0]['Age'],kde=True)
sns.displot(titanic[titanic['Survived']==1]['Age'],kde=True)
```

```python
sns.displot(
    data=titanic,
    x='Age',
    hue='Survived',
    kind='kde'
)
```

#5.HeatMap(Categorical-Categorical)

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

#6.ClusterMap(Categorical-Categorical)

```python
titanic.groupby('Pclass')['Survived'].mean()
sns.clustermap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```
