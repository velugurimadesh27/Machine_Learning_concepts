```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['has_covid']),df['has_covid'],test_size=0.2)
```

#1.Without column Transfer


```python
from sklearn.impute import SimpleImputer
```

```python
si = SimpleImputer()
X_train_fever = si.fit_transform(X_train[['fever']])
X_test_fever = si.fit_transform(X_test[['fever']])
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories=[['Mild','Strong']])
X_train_cough = oe.fit_transform(X_train[['cough']])
X_test_cough = oe.fit_transform(X_test[['cough']])
```

```python
X_train_cough.shape
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe  = OneHotEncoder(drop='first',sparse_output=True)
X_train_gender = ohe.fit_transform(X_train[['gender','city']]).toarray()
X_test_gender = ohe.fit_transform(X_test[['gender','city']]).toarray()
```

```python
X_train_gender.shape
```

```python
X_train_age = X_train.drop(columns=['gender','city','cough','fever']).values
X_test_age = X_test.drop(columns=['gender','city','cough','fever']).values
```

```python
X_train_age.shape
```

```python
X_train_concat = np.concatenate((X_train_age,X_train_fever,X_train_cough,X_train_gender),axis=1)
```

```python
X_test_concat = np.concatenate((X_test_age,X_test_fever,X_test_cough,X_test_gender),axis=1)
```

#With Column Transfer

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(transformers=[
    ('tnfe',SimpleImputer(),['fever']),
    ('tnf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    ('tnf3',OneHotEncoder(drop='first',sparse_output=False),['gender','city'])
],remainder='passthrough')
```

```python
transformer
```

```python
X_train_transform=transformer.fit_transform(X_train)
X_test_transform=transformer.fit_transform(X_test)
```
