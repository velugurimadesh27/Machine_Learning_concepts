```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/wine_data.csv',header=None,usecols=[0,1,2])
df.columns=['class_label','Alcohol','Malic_acid']
```

```python
df
```

```python
color_dict = {1:'red',2:'blue',3:'black'}
sns.scatterplot(x=df['Alcohol'],y=df['Malic_acid'],hue=df['class_label'],palette=color_dict)
```

```python
sns.kdeplot(df['Alcohol'])
```

```python
sns.kdeplot(df['Malic_acid'])
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df.drop('class_label',axis=1),df['class_label'],
                                               test_size=0.3,
                                               random_state=0
                                               )
```

```python
X_train
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
scaler.fit(X_train)
X_train_scaled=scaler.transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
X_train.mean()
```

```python
X_train_scaled.mean()
```

```python
X_train_scaled
```

```python
X_train_scaled=pd.DataFrame(X_train_scaled,columns=X_train.columns)
X_test_scaled=pd.DataFrame(X_test_scaled,columns=X_test.columns)
```

```python
X_train_scaled
```

```python
np.round(X_train.describe(),1)
```

```python
np.round(X_train_scaled.describe(),1)
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,figsize=(12,5))
ax1.scatter(X_train['Alcohol'],X_train['Malic_acid'])
ax1.set_title("Before scaling")
ax2.scatter(X_train_scaled['Alcohol'],X_train_scaled['Malic_acid'],color='Red')
ax2.set_title("Before scaling")
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,figsize=(12,5))
sns.kdeplot(X_train['Alcohol'],ax=ax1)
ax1.set_title("before scaling")
sns.kdeplot(X_train_scaled['Alcohol'],ax=ax2)
ax2.set_title("After scaling")
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,figsize=(12,5))
sns.kdeplot(X_train['Malic_acid'],ax=ax1)
ax1.set_title("before scaling")
sns.kdeplot(X_train_scaled['Malic_acid'],ax=ax2)
ax2.set_title("After scaling")
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,figsize=(12,5))
ax1.set_title('Before scaling')
sns.kdeplot(X_train['Alcohol'],ax=ax1)
sns.kdeplot(X_train['Malic_acid'],ax=ax1)
ax2.set_title('After scaling')
sns.kdeplot(X_train_scaled['Alcohol'],ax=ax2)
sns.kdeplot(X_train_scaled['Malic_acid'],ax=ax2)
```
